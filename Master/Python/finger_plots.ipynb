```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import func_notebook as nbk
from pathlib import Path
```

```python
areas= []
sigmas = []
bkg_avg = []

counter = 0
#apply the function to all the files in the folder
folder_path = Path('2026_07_15_LED')
#loop through all the subfolders
for subfolder in folder_path.iterdir():
    if subfolder.is_dir():
        #loop through all the files in the subfolder
        for file in subfolder.iterdir():
            if file.suffix == '.CSV' and file.name != 'INDEX.CSV':
                
                datapoints = nbk.compute_photon_signal_single_peak(file)
                print(f"Processing file: {file}", f", file number:",counter, f", datapoints:", datapoints)
                                
                if type(datapoints) == type(np.nan):
                    continue
                else:
                    areas.append(datapoints[0])
                    sigmas.append(datapoints[1])
                    bkg_avg.append(datapoints[2])
                    counter += 1

```

```python
plt.hist(areas, bins=400)
```
