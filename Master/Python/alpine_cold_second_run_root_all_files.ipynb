```python
%load_ext autoreload
%autoreload 2

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import func_notebook as nbk
from pathlib import Path
from scipy.signal import find_peaks
from scipy.ndimage import gaussian_filter1d
from itertools import islice
import pyarrow as py
import uproot
import awkward_pandas
import glob
```

```python

all_dataframes = []
counter = 1

# Update the glob search to look for your root files
file_list = glob.glob("/disk/gfs_atp/xlzd/alpine/run1_2/*/*.root") 

for filepath in file_list:

    print('*'*100)
    print('Processing file', filepath)
    print('File ', counter,'out of ', len(file_list), 'files')
    print('*'*100)
    counter +=1


    # 1. Open the ROOT file analyse all peaks and return the data as a df
    data = nbk.analyse_root_data(filepath)

    
    # 2. Run the fitting function on the obtained data set for this file
    fit_df = nbk.fit_all_channels_in_df(data, column_to_fit="peak_integral", plots='no')
    
    # 4. Extract metadata from the folder name
    metadata_dict = nbk.parse_folder_metadata(filepath)
    
    # 5. Add the extracted metadata as new columns
    for key, value in metadata_dict.items():
        fit_df.insert(0, key, value)
    
    # 6. Add to master list
    all_dataframes.append(fit_df)

# Combine everything!
master_results_df = pd.concat(all_dataframes, ignore_index=True)
master_results_df.to_parquet("master_fit_results.parquet", index=False)
```
